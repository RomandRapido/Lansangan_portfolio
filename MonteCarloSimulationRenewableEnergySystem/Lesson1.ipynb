```r
n <- 10000
x <- runif(n, -1, 1)
y <- runif(n, -1, 1)
inside_circle <- sum(x^2 + y^2 <= 1)
pi_estimate <- 4 * inside_circle / n
pi_estimate
```

```r
n <- 1000
k <- 0 # accepted counter
j <- 0 # iterations
y <- numeric(n) # accepted placeholder
c=1.5

while ( k < n){
    j <- j + 1
    u <- runif(1) # generate u
    
    x <- runif(1) # generate from g

    fx = 6*x*(1-x)

    if (u < fx/(c)){
        k <- k + 1
        y[k] <- x # note that R is one based.
    }
}

q <- qbeta(ppoints(y), 2,2)
qqplot(x=q, y=y, )
```
